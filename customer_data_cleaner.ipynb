```python
cust_data = [
    {"ID": 1, "name": "Shreya", "email": "shreya@example.com"},
    {"ID": 2, "name": "Amit", "email": ""},
    {"ID": 3, "name": "Neha", "email": "neha@example.com"},
    {"ID": 2, "name": "Amit", "email": ""}
]

list_of_seen_IDs = []

for cust in cust_data:
    if cust["ID"] not in list_of_seen_IDs:
        list_of_seen_IDs.append(cust["ID"])
    else:
        print(f'Duplicate ID found: {cust["ID"]}')

    if cust["email"] == "":
        print(f'Missing email: {cust["name"]}')
```

```python
cust_data = [
    {"ID": 1, "name": "Shreya", "email": "shreya@example.com"},
    {"ID": 2, "name": "Amit", "email": ""},
    {"ID": 3, "name": "Neha", "email": "neha@example.com"},
    {"ID": 2, "name": "Amit", "email": ""}
]

valid_customers = []
invalid_customers = []
seen_ids = []

for cust in cust_data:

    if cust["ID"] in seen_ids or cust["email"] == "":
        invalid_customers.append(cust)

        if cust["ID"] in seen_ids:
            print(f'Duplicate ID found: {cust["ID"]}')

        if cust["email"] == "":
            print(f'Missing email: {cust["name"]}')

    else:
        valid_customers.append(cust)

    if cust["ID"] not in seen_ids:
        seen_ids.append(cust["ID"])

print("Valid customers:", valid_customers)
print("Invalid customers:", invalid_customers)
print("Valid customers:", len(valid_customers))
print("Invalid customers:", len(invalid_customers))
```

```python

import csv

# Save valid customers
with open("valid_customers.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["ID", "name", "email"])
    writer.writeheader()
    writer.writerows(valid_customers)

# Save invalid customers
with open("invalid_customers.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["ID", "name", "email"])
    writer.writeheader()
    writer.writerows(invalid_customers)

print("Customer data saved successfully!")
```
